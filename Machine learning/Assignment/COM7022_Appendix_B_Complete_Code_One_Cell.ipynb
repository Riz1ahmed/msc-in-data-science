```python
# * When run, If the 'Sales-[5444].csv' file is not present, the code will show
#   a Colab upload prompt.



#============================================================================
# 01. Environment and reproducibility
# Import numerical, machine-learning, and plotting libraries; set a fixed random seed.
# Generated figures and tables are saved to COM7022_outputs/.
#============================================================================

from pathlib import Path
import re
import sys
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import sklearn
import scipy

from IPython.display import display
from sklearn.cluster import AgglomerativeClustering, DBSCAN, KMeans
from sklearn.compose import ColumnTransformer
from sklearn.decomposition import PCA
from sklearn.impute import SimpleImputer
from sklearn.metrics import (
    silhouette_score, davies_bouldin_score, calinski_harabasz_score,
    adjusted_rand_score
)
from sklearn.neighbors import NearestNeighbors
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import FunctionTransformer, OneHotEncoder, StandardScaler
from scipy.cluster.hierarchy import linkage, dendrogram

RANDOM_STATE = 42
np.random.seed(RANDOM_STATE)
sns.set_theme(style="whitegrid", palette="deep")
plt.rcParams.update({"figure.dpi": 110, "savefig.dpi": 170, "axes.titlesize": 12})

OUTPUT_DIR = Path("COM7022_outputs")
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

print("Python:", sys.version.split()[0])
print("pandas:", pd.__version__, "| NumPy:", np.__version__)
print("scikit-learn:", sklearn.__version__, "| SciPy:", scipy.__version__)
print("Random seed:", RANDOM_STATE)
print("Output folder:", OUTPUT_DIR.resolve())

def save_fig(filename):
    """Save the current Matplotlib figure as a high-resolution PNG."""
    plt.savefig(OUTPUT_DIR / filename, dpi=170, bbox_inches="tight")
    plt.show()


#============================================================================
# 02. Load and validate the source dataset
# a. Use the supplied Sales-[5444].csv file. The code accepts the Colab Files panel or
#    prompts for a file upload if it cannot find the CSV.
# b. Check all required columns.
#============================================================================


DATASET_NAME = "Sales-[5444].csv"
candidates = [Path("/content") / DATASET_NAME,
              Path.cwd() / DATASET_NAME,
              Path("/mnt/data") / DATASET_NAME]
source_path = next((p for p in candidates if p.is_file()), None)

if source_path is None:
    try:
        from google.colab import files
        print("Dataset not found. Please upload Sales-[5444].csv in the popup.")
        uploaded = files.upload()
        if DATASET_NAME not in uploaded:
            raise FileNotFoundError(f"Please upload the exact file: {DATASET_NAME}")
        source_path = Path(DATASET_NAME)
    except ImportError as exc:
        raise FileNotFoundError(
            f"Could not find {DATASET_NAME}. Put it in the same folder as the notebook."
        ) from exc

raw = pd.read_csv(source_path)
required_cols = {
    "Item_Name", "details", "Catatory", "price", "priceWithCurrency",
    "available", "text", "sold", "lastUpdated", "itemLocation"
}
if not required_cols.issubset(raw.columns):
    raise ValueError("Unexpected dataset columns. Missing: " +
                     str(sorted(required_cols - set(raw.columns))))

print("Dataset loaded from:", source_path)
print("Shape (rows, columns):", raw.shape)
print("Columns:", raw.columns.tolist())
display(raw.head(5))


#============================================================================
# 03. Data audit and descriptive statistics
# Check data types, missing values, distinct values, duplicates and the distribution
# of price, items sold and availability before making cleaning decisions.
#============================================================================

audit = pd.DataFrame({
    "dtype": raw.dtypes.astype(str),
    "missing_n": raw.isna().sum(),
    "missing_pct": (100 * raw.isna().mean()).round(2),
    "unique_non_null": raw.nunique(dropna=True),
})
print("Exact duplicate listing rows:", int(raw.duplicated().sum()))
print("Data types and missingness:")
display(audit)
display(raw[["price", "sold", "available"]].describe().round(2))
display(raw["Catatory"].value_counts(dropna=False).head(12).to_frame("listings"))
audit.to_csv(OUTPUT_DIR / "01_original_data_audit.csv", index_label="feature")


#============================================================================
# 04. Visual audit of the raw data
# Plot initial missingness and numeric distributions. Retain these plots as evidence
# of why preprocessing was required.
#============================================================================

fig, axes = plt.subplots(1, 2, figsize=(13, 4.2))
audit["missing_n"].sort_values(ascending=False).plot.bar(ax=axes[0], color="#4776a8")
axes[0].set(title="Missing values by original feature", ylabel="Number of rows", xlabel="Feature")
axes[0].tick_params(axis="x", rotation=55)

for col, colour in zip(["price", "sold", "available"], ["#4776a8", "#c07065", "#4b8b84"]):
    vals = pd.to_numeric(raw[col], errors="coerce").dropna().clip(lower=0)
    sns.kdeplot(x=np.log1p(vals).to_numpy(), ax=axes[1], label=col, color=colour, fill=False, warn_singular=False)
axes[1].set(title="Numeric distributions after log(1 + value) (visual only)",
            xlabel="log(1 + original value)")
axes[1].legend()
fig.tight_layout()
save_fig("02_missingness_and_numeric_distributions.png")


#============================================================================
# 05. Data cleaning and category standardisation
# Recover only explicitly stated USD prices or sold/stock counts from text; do not
# invent numbers for vague availability phrases. Identify CAD labels and exclude
# those rows from USD-based modelling. Group category variants consistently.
#============================================================================

cleaned = raw.copy()
cleaning_log = []

def log_change(field, rule, count):
    cleaning_log.append({"field": field, "action": rule, "records_affected": int(count)})

# Standardise numeric types without blindly converting currencies.
for col in ["price", "sold", "available"]:
    cleaned[col] = pd.to_numeric(cleaned[col], errors="coerce")

currency_text = cleaned["priceWithCurrency"].fillna("").astype(str).str.strip()
cleaned["currency_status"] = np.select(
    [currency_text.str.match(r"(?i)^US\s*\$"),
     currency_text.str.match(r"(?i)^C\s*\$")],
    ["USD", "CAD"], default="Unknown"
)

# Extract prices only where an explicit US-dollar label is present.
usd_price = currency_text.str.extract(r"(?i)^US\s*\$\s*([\d,]+(?:\.\d+)?)")[0]
usd_price = pd.to_numeric(usd_price.str.replace(",", "", regex=False), errors="coerce")
price_mask = cleaned["price"].isna() & usd_price.notna()
cleaned.loc[price_mask, "price"] = usd_price.loc[price_mask]
log_change("price", "Recovered from explicitly USD price text", price_mask.sum())

stock_text = cleaned["text"].fillna("").astype(str)
sold_from_text = stock_text.str.extract(r"(?i)([\d,]+)\s+sold\b")[0]
sold_from_text = pd.to_numeric(sold_from_text.str.replace(",", "", regex=False), errors="coerce")
sold_mask = cleaned["sold"].isna() & sold_from_text.notna()
cleaned.loc[sold_mask, "sold"] = sold_from_text.loc[sold_mask]
log_change("sold", "Recovered explicit sold count from stock text", sold_mask.sum())

# Recover only an explicit stock count, or the phrase 'Last One'.
exact_available = stock_text.str.extract(r"(?i)^\s*([\d,]+)\s+available\b")[0]
exact_available = pd.to_numeric(exact_available.str.replace(",", "", regex=False), errors="coerce")
last_one = stock_text.str.contains(r"(?i)^\s*last\s+one\b", regex=True, na=False)
avail_mask = cleaned["available"].isna() & exact_available.notna()
cleaned.loc[avail_mask, "available"] = exact_available.loc[avail_mask]
last_one_mask = cleaned["available"].isna() & last_one
cleaned.loc[last_one_mask, "available"] = 1.0
log_change("available", "Recovered explicitly stated exact N available", avail_mask.sum())
log_change("available", "Recovered Last One as 1 available", last_one_mask.sum())

cleaned["availability_censored_text"] = stock_text.str.contains(
    r"(?i)more than\s+\d+\s+available|more than\s+\d+\s+lots\s+available",
    regex=True, na=False
)
cleaned["availability_unresolved"] = cleaned["available"].isna()

# Preserve the original category while mapping common variants to stable groups.
def normalise_category(v):
    if pd.isna(v):
        return "Other / Unclear"
    s = re.sub(r"\s+", " ", str(v).strip().casefold())
    has_edp = bool(re.search(r"eau\s+de\s+parf(?:um|ume)|\bedp\b", s))
    has_edt = bool(re.search(r"eau\s+de\s+toil(?:ette|lette)|\bedt\b", s))
    has_edc = bool(re.search(r"eau\s+de\s+cologne|\bedc\b|\bcologne\b", s))
    if sum([has_edp, has_edt, has_edc]) > 1:
        return "Mixed Concentrations"
    if has_edp:
        return "Eau de Parfum"
    if has_edt:
        return "Eau de Toilette"
    if has_edc:
        return "Cologne"
    if "gift" in s or "set" in s:
        return "Gift Set"
    if "parfum" in s or "perfume" in s or "extrait" in s or "elixir" in s:
        return "Perfume / Parfum"
    if "spray" in s or "oil" in s or "roll" in s or "aftershave" in s or "deodorant" in s:
        return "Body Care / Oil"
    if "fragrance" in s or "pheromone" in s:
        return "General Fragrance"
    return "Other / Unclear"

cleaned["category_group"] = cleaned["Catatory"].apply(normalise_category)
log_change("Catatory", "Standardised into product category groups", len(cleaned))

# Exclude clearly CAD-priced rows from a USD-based comparison, NOT entire dataset.
cad_mask = cleaned["currency_status"].eq("CAD")
model_data = cleaned.loc[~cad_mask].copy().reset_index(drop=True)
log_change("priceWithCurrency", "Excluded explicit CAD prices from USD-only modelling", cad_mask.sum())

# Ignore nonsensical numeric negatives, if any; preserve and report count.
for col in ["price", "sold", "available"]:
    negative = model_data[col].lt(0)
    log_change(col, "Negative values set to missing for median imputation", negative.sum())
    model_data.loc[negative, col] = np.nan

# Record quality flags (not included in the distance matrix).
for col in ["price", "sold", "available"]:
    model_data[f"missing_{col}_at_model_time"] = model_data[col].isna()

log_table = pd.DataFrame(cleaning_log)
missing_comparison = pd.DataFrame({
    "original_missing": raw[["price", "sold", "available"]].isna().sum(),
    "after_targeted_recovery_in_modelling_rows": model_data[["price", "sold", "available"]].isna().sum()
})
print("Original rows:", len(raw), "| Modelling rows:", len(model_data))
print("Currency labels:")
display(cleaned["currency_status"].value_counts(dropna=False).to_frame("rows"))
print("Cleaning log:")
display(log_table)
print("Missing values before vs after targeted recovery (note USD-only subset):")
display(missing_comparison)
print("Normalised category counts:")
display(model_data["category_group"].value_counts().to_frame("rows"))

log_table.to_csv(OUTPUT_DIR / "03_cleaning_log.csv", index=False)
missing_comparison.to_csv(OUTPUT_DIR / "04_missingness_comparison.csv")


#============================================================================
# 06. Visual checks after cleaning
# Compare missingness before and after targeted recovery; show the distribution
# of standardised product categories.
#============================================================================

fig, axes = plt.subplots(1, 2, figsize=(12.8, 4))
missing_comparison.plot(kind="bar", ax=axes[0], color=["#c07065", "#4b8b84"])
axes[0].set(title="Missing values: before versus targeted recovery", xlabel="Numeric feature", ylabel="Rows")
axes[0].tick_params(axis="x", rotation=0)
model_data["category_group"].value_counts().plot.barh(ax=axes[1], color="#497fa7")
axes[1].invert_yaxis()
axes[1].set(title="Standardised product category distribution", xlabel="Rows", ylabel="Category")
fig.tight_layout()
save_fig("05_cleaning_and_categories.png")


#============================================================================
# 07. Feature preparation and scaling
# Use price, sold, available and standardised product category as clustering inputs.
# Apply median imputation, log1p, standard scaling and one-hot encoding.
# Unresolved missingness is handled inside the preprocessing pipeline.
#============================================================================

numeric_features = ["price", "sold", "available"]
categorical_features = ["category_group"]
feature_cols = numeric_features + categorical_features

# Make log1p safe even if future data contains small negative errors: negatives are NaN above.
def log1p_nonnegative(X):
    return np.log1p(np.maximum(np.asarray(X, dtype=float), 0.0))

num_pipeline = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("log1p", FunctionTransformer(log1p_nonnegative, validate=False)),
    ("scaler", StandardScaler())
])
cat_pipeline = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore", sparse_output=False))
])
preprocessor = ColumnTransformer(transformers=[
    ("numeric", num_pipeline, numeric_features),
    ("category", cat_pipeline, categorical_features)
], remainder="drop", sparse_threshold=0.0)

X = np.asarray(preprocessor.fit_transform(model_data[feature_cols]), dtype=float)
encoded_names = preprocessor.named_transformers_["category"].named_steps["onehot"].get_feature_names_out(categorical_features).tolist()
feature_names = numeric_features + encoded_names

assert X.ndim == 2 and len(X) == len(model_data)
assert np.isfinite(X).all(), "Transformed features contain inf or NaN. Inspect cleaning."
assert X.shape[1] >= 4, "Unexpected feature preprocessing result"

print("Modelling rows:", len(model_data))
print("Selected raw features:", feature_cols)
print("Transformed matrix shape:", X.shape)
print("Transformed names:", feature_names)
display(pd.DataFrame(X[:5], columns=feature_names).round(3))

# For descriptive profiling only, not the model input: imputed numeric values.
for col in numeric_features:
    model_data[f"{col}_profile"] = model_data[col].fillna(model_data[col].median())

pd.DataFrame({"transformed_feature": feature_names}).to_csv(
    OUTPUT_DIR / "06_encoded_features.csv", index=False)


#============================================================================
# 08. Consistent, valid clustering metrics
# Define Silhouette (higher is better), Davies-Bouldin (lower is better) and
# Calinski-Harabasz (higher is better). DBSCAN noise points are excluded from
# its internal metrics; coverage is reported separately.
#============================================================================

def score_clustering(X_input, labels, method_name, params):
    """Calculate only metrics valid for this clustering; DBSCAN noise excluded."""
    labels = np.asarray(labels, dtype=int)
    assigned = labels != -1
    n_total = len(labels)
    n_assigned = int(assigned.sum())
    use_X = X_input[assigned]
    use_labels = labels[assigned]
    label_counts = pd.Series(use_labels).value_counts()
    n_clusters = int(label_counts.size)
    result = {
        "model": method_name, "parameters": str(params),
        "n_clusters": n_clusters, "assigned_n": n_assigned,
        "coverage_pct": round(100 * n_assigned / n_total, 2),
        "noise_pct": round(100 * (n_total - n_assigned) / n_total, 2),
        "min_cluster_n": int(label_counts.min()) if n_clusters else 0,
        "largest_cluster_pct_assigned": round(100 * label_counts.max() / n_assigned, 2)
        if n_clusters and n_assigned else np.nan,
        "silhouette": np.nan, "davies_bouldin": np.nan,
        "calinski_harabasz": np.nan
    }
    if 2 <= n_clusters < n_assigned:
        try:
            result["silhouette"] = float(silhouette_score(use_X, use_labels))
            result["davies_bouldin"] = float(davies_bouldin_score(use_X, use_labels))
            result["calinski_harabasz"] = float(calinski_harabasz_score(use_X, use_labels))
        except ValueError as exc:
            warnings.warn(f"Metrics undefined for {method_name} {params}: {exc}")
    return result


def choose_candidate(scores, require_coverage=0):
    """Prefer usable segments; if none, report best valid candidate and warn."""
    t = scores.dropna(subset=["silhouette"]).copy()
    if t.empty:
        return None, "No candidate with 2+ measurable clusters"
    viable = t[
        (t["coverage_pct"] >= require_coverage) &
        (t["largest_cluster_pct_assigned"] <= 90) &
        (t["min_cluster_n"] >= np.maximum(5, np.ceil(0.02 * t["assigned_n"])))
    ]
    if not viable.empty:
        return viable.sort_values("silhouette", ascending=False).iloc[0], "business-size safeguards met"
    return t.sort_values("silhouette", ascending=False).iloc[0], "fallback: assess imbalance/noise manually"


#============================================================================
# 09. K-Means: parameter search
# Test k=2 through k=8 and select a candidate using both separation quality and
# cluster-size safeguards, rather than the maximum Silhouette Score alone.
#============================================================================

K_RANGE = list(range(2, 9))
kmeans_results = []
kmeans_labels_by_k = {}
for k in K_RANGE:
    km = KMeans(n_clusters=k, n_init=20, random_state=RANDOM_STATE)
    lab = km.fit_predict(X)
    rec = score_clustering(X, lab, "K-Means", {"k": k, "n_init": 20})
    rec["k"] = k
    rec["inertia"] = float(km.inertia_)
    kmeans_results.append(rec)
    kmeans_labels_by_k[k] = lab

kmeans_grid = pd.DataFrame(kmeans_results)
kmeans_best, kmeans_selection_note = choose_candidate(kmeans_grid, require_coverage=100)
if kmeans_best is None:
    raise RuntimeError("No K-Means candidate yielded valid clustering metrics")
selected_k = int(kmeans_best["k"])
labels_kmeans = kmeans_labels_by_k[selected_k]
print("Selected K-Means k:", selected_k, "|", kmeans_selection_note)
display(kmeans_grid[["k", "inertia", "silhouette", "davies_bouldin",
                     "calinski_harabasz", "min_cluster_n", "largest_cluster_pct_assigned"]].round(3))
kmeans_grid.to_csv(OUTPUT_DIR / "08_kmeans_parameter_search.csv", index=False)


#============================================================================
# 10. K-Means: elbow and Silhouette plots
# Draw inertia and Silhouette Score against k. These figures support the
# interpretation of candidate cluster counts.
#============================================================================

fig, axes = plt.subplots(1, 2, figsize=(12, 4.2))
axes[0].plot(kmeans_grid["k"], kmeans_grid["inertia"], marker="o")
axes[0].set(title="K-Means elbow: inertia versus k", xlabel="k", ylabel="Within-cluster inertia")
axes[1].plot(kmeans_grid["k"], kmeans_grid["silhouette"], marker="o")
axes[1].axvline(selected_k, linestyle="--", color="#bd5b48", label=f"Selected k = {selected_k}")
axes[1].set(title="K-Means silhouette versus k", xlabel="k", ylabel="Silhouette (higher is better)")
axes[1].legend()
fig.tight_layout()
save_fig("09_kmeans_elbow_silhouette.png")


#============================================================================
# 11. Hierarchical clustering: linkage search
# Compare Ward, complete and average linkage over k=2 through k=8.
# Penalise overly imbalanced partitions through explicit size safeguards.
#============================================================================

HIER_LINKAGES = ["ward", "complete", "average"]
hier_results, hier_labels = [], {}
for how in HIER_LINKAGES:
    for k in K_RANGE:
        ag = AgglomerativeClustering(n_clusters=k, linkage=how)
        lab = ag.fit_predict(X)
        rec = score_clustering(X, lab, "Agglomerative", {"linkage": how, "k": k})
        rec["linkage"] = how
        rec["k"] = k
        hier_results.append(rec)
        hier_labels[(how, k)] = lab
hier_grid = pd.DataFrame(hier_results)
hier_best, hier_selection_note = choose_candidate(hier_grid, require_coverage=100)
if hier_best is None:
    raise RuntimeError("No hierarchical candidate yielded valid clustering metrics")
chosen_linkage = hier_best["linkage"]
chosen_hier_k = int(hier_best["k"])
labels_hier = hier_labels[(chosen_linkage, chosen_hier_k)]
print("Selected Agglomerative:", chosen_linkage, "linkage; k =", chosen_hier_k)
print("Selection note:", hier_selection_note)
display(hier_grid.sort_values("silhouette", ascending=False)[
    ["linkage", "k", "silhouette", "davies_bouldin",
     "min_cluster_n", "largest_cluster_pct_assigned"]].head(12).round(3))
hier_grid.to_csv(OUTPUT_DIR / "10_hierarchical_parameter_search.csv", index=False)


#============================================================================
# 12. Hierarchical clustering: sample dendrogram
# Use a reproducible sample only for an illustrative dendrogram.
# The full dataset is used for fitting and comparison.
#============================================================================

# Interpretability: a fixed random sample, not a dendrogram of the complete model.
sample_n = min(130, len(X))
sample_indices = np.random.default_rng(RANDOM_STATE).choice(len(X), size=sample_n, replace=False)
Z = linkage(X[sample_indices], method=chosen_linkage, metric="euclidean")
plt.figure(figsize=(12.5, 4.3))
dendrogram(Z, truncate_mode="lastp", p=25, leaf_rotation=45,
           show_leaf_counts=True, no_labels=False)
plt.title(f"Illustrative dendrogram (sample n={sample_n}, linkage={chosen_linkage})")
plt.xlabel("Merged sample subtrees / observation indices")
plt.ylabel("Euclidean linkage distance")
plt.tight_layout()
save_fig("11_hierarchical_sample_dendrogram.png")


#============================================================================
# 13. DBSCAN: nearest-neighbour diagnostics
# Plot sorted nearest-neighbour distances as guidance for eps candidates.
# The plot is diagnostic, not proof that one eps is globally optimal.
#============================================================================

# The 'k-distance' plot is a diagnostic only, not proof of an optimum epsilon.
NEIGHBOUR_OPTIONS = [5, 10, 15]
quantiles = [0.45, 0.55, 0.65, 0.75, 0.82, 0.88, 0.92, 0.95, 0.98]
knn_distance_by_k = {}
for n_neighbors in NEIGHBOUR_OPTIONS:
    neigh = NearestNeighbors(n_neighbors=n_neighbors, metric="euclidean").fit(X)
    dists, _ = neigh.kneighbors(X)
    knn_distance_by_k[n_neighbors] = np.sort(dists[:, -1])

fig, ax = plt.subplots(figsize=(10, 4.2))
for k in NEIGHBOUR_OPTIONS:
    ax.plot(knn_distance_by_k[k], label=f"{k}-neighbour distance")
ax.set(title="Sorted k-nearest-neighbour distances (DBSCAN guidance)",
       xlabel="Observation index, sorted by kth distance", ylabel="Euclidean distance")
ax.legend()
fig.tight_layout()
save_fig("12_dbscan_knn_distance.png")


#============================================================================
# 14. DBSCAN: parameter search
# Test eps and min_samples combinations. Calculate the fraction assigned to
# clusters, the fraction marked noise and valid internal evaluation metrics.
#============================================================================

dbscan_results, dbscan_labels = [], {}
for min_samples in NEIGHBOUR_OPTIONS:
    eps_candidates = np.unique(np.round(np.quantile(
        knn_distance_by_k[min_samples], quantiles), 5))
    for eps in eps_candidates:
        db = DBSCAN(eps=float(eps), min_samples=min_samples, metric="euclidean")
        lab = db.fit_predict(X)
        rec = score_clustering(X, lab, "DBSCAN", {"eps": float(eps), "min_samples": min_samples})
        rec["eps"] = float(eps)
        rec["min_samples"] = min_samples
        dbscan_results.append(rec)
        dbscan_labels[(float(eps), min_samples)] = lab

dbscan_grid = pd.DataFrame(dbscan_results)
dbscan_best, dbscan_selection_note = choose_candidate(dbscan_grid, require_coverage=65)
if dbscan_best is not None:
    db_eps = float(dbscan_best["eps"])
    db_min_samples = int(dbscan_best["min_samples"])
    labels_dbscan = dbscan_labels[(db_eps, db_min_samples)]
    print("Selected DBSCAN eps:", db_eps, "min_samples:", db_min_samples)
    print("DBSCAN selection note:", dbscan_selection_note)
else:
    db_eps, db_min_samples = np.nan, None
    labels_dbscan = np.full(len(X), -1, dtype=int)
    print("No DBSCAN setting had 2+ meaningful groups: reporting result as unsuitable.")

print("Top DBSCAN choices with valid internal metrics:")
display(dbscan_grid.sort_values("silhouette", ascending=False, na_position="last")[[
    "eps", "min_samples", "n_clusters", "noise_pct", "coverage_pct",
    "silhouette", "davies_bouldin", "min_cluster_n",
    "largest_cluster_pct_assigned"]].head(12).round(3))
dbscan_grid.to_csv(OUTPUT_DIR / "13_dbscan_parameter_search.csv", index=False)


#============================================================================
# 15. Compare selected candidate models
# Present all three methods in a comparison table. The final suggested model
# uses interpretability/coverage safeguards in addition to Silhouette Score.
#============================================================================

selected_rows = [kmeans_best.to_dict(), hier_best.to_dict()]
if dbscan_best is not None:
    selected_rows.append(dbscan_best.to_dict())
else:
    selected_rows.append(score_clustering(X, labels_dbscan, "DBSCAN", "no suitable candidate"))

comparison = pd.DataFrame(selected_rows)
comparison["model"] = ["K-Means", "Agglomerative", "DBSCAN"]
comparison_columns = ["model", "parameters", "n_clusters", "assigned_n", "coverage_pct",
                      "noise_pct", "min_cluster_n", "largest_cluster_pct_assigned",
                      "silhouette", "davies_bouldin", "calinski_harabasz"]
comparison = comparison[comparison_columns].copy()

# Explicit and adjustable business constraints, not a claimed true optimiser.
viable = comparison.loc[
    comparison["silhouette"].notna() &
    (comparison["coverage_pct"] >= 85) &
    (comparison["n_clusters"].between(2, 8)) &
    (comparison["largest_cluster_pct_assigned"] <= 90) &
    (comparison["min_cluster_n"] >= np.maximum(5, np.ceil(0.02 * comparison["assigned_n"])))
]
if viable.empty:
    # A full-coverage model makes the comparison interpretable if safeguards fail.
    viable = comparison.loc[
        comparison["model"].isin(["K-Means", "Agglomerative"]) &
        comparison["silhouette"].notna()
    ]
    print("Warning: No model met all safeguards; recommendation is provisional.")
if viable.empty:
    raise RuntimeError("No usable clustering models available. Inspect parameter searches.")

suggested_model = str(viable.sort_values("silhouette", ascending=False).iloc[0]["model"])
model_labels = {"K-Means": labels_kmeans, "Agglomerative": labels_hier,
                "DBSCAN": labels_dbscan}
labels_final = model_labels[suggested_model]

print("MODEL COMPARISON (actual computed results)")
display(comparison.round(3))
print("Suggested model for further BUSINESS review:", suggested_model)
print("Interpretation note: this is a preliminary, evidence-based recommendation, not a causal result.")
comparison.to_csv(OUTPUT_DIR / "14_model_comparison.csv", index=False)


#============================================================================
# 16. Visual comparison of model scores
# Show Silhouette Score and data coverage together; avoid treating the DBSCAN
# internal score as directly equivalent to full-coverage model scores.
#============================================================================

fig, axes = plt.subplots(1, 2, figsize=(11.8, 4.1))
comparison.plot.bar(x="model", y="silhouette", ax=axes[0], legend=False, color="#497fa7")
axes[0].set(title="Silhouette score (DBSCAN excludes noise)", xlabel="", ylabel="Higher is better")
axes[0].tick_params(axis="x", rotation=0)
comparison.plot.bar(x="model", y="coverage_pct", ax=axes[1], legend=False, color="#4b8b84")
axes[1].axhline(100, linestyle="--", color="grey", alpha=0.7)
axes[1].set(title="Percentage of product listings assigned", xlabel="", ylabel="Coverage (%)", ylim=(0, 105))
axes[1].tick_params(axis="x", rotation=0)
fig.tight_layout()
save_fig("15_comparison_scores_coverage.png")


#============================================================================
# 17. PCA: dimensionality reduction and visual comparison
# Project the encoded, scaled data into two principal components for display.
# Report variance explained; do not treat the 2D plot as proof of clustering quality.
#============================================================================

pca = PCA(n_components=2, random_state=RANDOM_STATE)
X_2d = pca.fit_transform(X)
pca_variance = pca.explained_variance_ratio_
print("PC1 variance explained:", f"{pca_variance[0]:.2%}")
print("PC2 variance explained:", f"{pca_variance[1]:.2%}")
print("Combined 2D variance explained:", f"{pca_variance.sum():.2%}")

pca_view = pd.DataFrame({"PC1": X_2d[:, 0], "PC2": X_2d[:, 1],
                         "kmeans_cluster": labels_kmeans,
                         "hierarchical_cluster": labels_hier,
                         "dbscan_cluster": labels_dbscan})
pca_view.to_csv(OUTPUT_DIR / "16_pca_coordinates_and_labels.csv", index=False)

fig, axes = plt.subplots(1, 3, figsize=(16, 4.4), sharex=True, sharey=True)
for ax, (name, labels) in zip(axes, model_labels.items()):
    labels = np.asarray(labels)
    valid_cluster_ids = sorted(np.unique(labels[labels >= 0]).tolist())
    for c in valid_cluster_ids:
        mask = labels == c
        ax.scatter(X_2d[mask, 0], X_2d[mask, 1], s=11, alpha=0.6, label=f"C{c}")
    noise_mask = labels == -1
    if np.any(noise_mask):
        ax.scatter(X_2d[noise_mask, 0], X_2d[noise_mask, 1],
                   s=13, c="grey", marker="x", alpha=0.4, label="Noise")
    ax.set(title=f"{name} | clusters={len(valid_cluster_ids)}", xlabel="PC1")
    ax.legend(title="Group", fontsize=7, title_fontsize=8, ncol=2, loc="best")
axes[0].set_ylabel("PC2")
fig.suptitle(f"PCA visualisation (PC1 + PC2 explain {pca_variance.sum():.1%} of variance)")
fig.tight_layout()
save_fig("17_pca_clustering_three_models.png")


#============================================================================
# 18. Selected-cluster profiles
# Summarise cluster sizes, typical price, cumulative sales count, availability
# and common categories in original interpretable units.
#============================================================================

profile_df = model_data.copy()
profile_df["cluster"] = labels_final
profile_df["cluster_name"] = profile_df["cluster"].map(lambda z: "Noise" if z == -1 else f"Cluster {z}")

def dominant_category(s):
    if s.empty:
        return "N/A"
    return str(s.value_counts(dropna=False).index[0])

profile = (profile_df.groupby("cluster_name", as_index=False)
    .agg(listings=("price", "size"),
         median_price_usd=("price", "median"),
         median_sold_count=("sold", "median"),
         median_reported_available=("available", "median"),
         missing_availability_pct=("available", lambda s: 100 * s.isna().mean()),
         most_common_category=("category_group", dominant_category)))
profile["pct_of_modelled_listings"] = (100 * profile["listings"] / len(profile_df)).round(2)
profile = profile.sort_values("listings", ascending=False).reset_index(drop=True)
print("Selected model:", suggested_model)
display(profile.round(2))

# Show relative category mix to make group interpretation less arbitrary.
category_mix = pd.crosstab(profile_df["cluster_name"],
                           profile_df["category_group"], normalize="index") * 100
print("Within-cluster category mix (%):")
display(category_mix.round(1))

profile.to_csv(OUTPUT_DIR / "18_final_cluster_profile.csv", index=False)
category_mix.to_csv(OUTPUT_DIR / "19_final_category_mix_pct.csv")


#============================================================================
# 19. Segment comparison charts
# Plot group size, price and sales to help interpret the selected segmentation
# for a non-technical business audience.
#============================================================================

plot_profile = profile.loc[profile["cluster_name"] != "Noise"].copy()
fig, axes = plt.subplots(1, 3, figsize=(15, 4.5))
for ax, y, ttl in zip(
    axes,
    ["listings", "median_price_usd", "median_sold_count"],
    ["Products per assigned cluster", "Median listing price (USD)", "Median cumulative sold count"]
):
    sns.barplot(data=plot_profile, x="cluster_name", y=y, ax=ax, color="#4779a5")
    ax.set(title=ttl, xlabel="Cluster", ylabel=y.replace("_", " "))
    ax.tick_params(axis="x", rotation=35)
fig.tight_layout()
save_fig("20_cluster_size_price_sales.png")


#============================================================================
# 20. Price and sales scatter plot
# Check whether selected groups differ in the observed price-sales relationship.
# Clusters describe associations and do not prove causal effects.
#============================================================================

# Descriptive 2D relationship; colours denote clusters, not supervised labels.
fig, ax = plt.subplots(figsize=(10, 5))
for label, subset in profile_df.groupby("cluster_name"):
    ax.scatter(np.log1p(subset["price"]), np.log1p(subset["sold"]),
               s=22, alpha=0.48, label=label)
ax.set(title=f"Price versus sales by {suggested_model} cluster (descriptive)",
       xlabel="log(1 + product price in USD)", ylabel="log(1 + cumulative sold count)")
ax.legend(title="Cluster", fontsize=8, bbox_to_anchor=(1.02, 1), loc="upper left")
fig.tight_layout()
save_fig("21_price_vs_sold_by_cluster.png")


#============================================================================
# 21. Cautious business implications
# Generate first-pass cluster descriptions and possible business actions.
# Recommendations require verification with time-based commercial information.
#============================================================================

ref_price = float(profile_df["price"].median())
ref_sold = float(profile_df["sold"].median())
ref_stock = float(profile_df["available"].median())

business_notes = []
for row in profile.itertuples(index=False):
    if row.cluster_name == "Noise":
        insight = "Unassigned DBSCAN records: audit atypical listings; not a coherent segment."
        action = "Review anomalies individually before any operational action."
    else:
        price_level = "higher-priced" if pd.notna(row.median_price_usd) and row.median_price_usd > ref_price else "lower-priced"
        sales_level = "higher cumulative sales" if pd.notna(row.median_sold_count) and row.median_sold_count > ref_sold else "lower cumulative sales"
        insight = (f"A {price_level} group with {sales_level}; "
                   f"main category: {row.most_common_category}.")
        if pd.notna(row.median_sold_count) and row.median_sold_count > ref_sold:
            action = "Check demand and available stock; prioritise product-level replenishment review."
        else:
            action = "Review price positioning, merchandising and slow-moving inventory; validate with time-based data."
    business_notes.append({"segment": row.cluster_name, "evidence_based_description": insight,
                           "potential_action_to_validate": action})

business_notes = pd.DataFrame(business_notes)
display(business_notes)
business_notes.to_csv(OUTPUT_DIR / "22_preliminary_business_notes.csv", index=False)
print("Dataset reference medians: price USD", round(ref_price, 2),
      "; sold", round(ref_sold, 2), "; reported availability", round(ref_stock, 2))


#============================================================================
# 22. Reproducibility and sensitivity check
# Refit K-Means with multiple fixed seeds and calculate Adjusted Rand Index
# relative to the reference assignment. This tests initialisation sensitivity.
#============================================================================

seed_list = [7, 21, 42, 99, 2026]
stability_records = []
for s in seed_list:
    other = KMeans(n_clusters=selected_k, n_init=20, random_state=s).fit_predict(X)
    stability_records.append({"seed": s, "ARI_vs_seed_42": adjusted_rand_score(labels_kmeans, other)})
stability_table = pd.DataFrame(stability_records)
display(stability_table.round(4))
print("Mean ARI across these seeds vs reference:",
      round(float(stability_table["ARI_vs_seed_42"].mean()), 4))
stability_table.to_csv(OUTPUT_DIR / "23_kmeans_seed_stability.csv", index=False)


#============================================================================
# 23. Save results and traceable assignments
# Export item-to-cluster assignments, summary text, and supporting CSV/PNG files.
# A separate written report and genuine Colab screenshots are still required.
#============================================================================

# Save a tidy record of every listing's selected cluster (for traceability).
export_cols = ["Item_Name", "Catatory", "category_group", "price", "sold",
               "available", "currency_status", "itemLocation"]
cluster_assignments = profile_df[export_cols + ["cluster", "cluster_name"]].copy()
cluster_assignments.to_csv(OUTPUT_DIR / "24_selected_model_cluster_assignments.csv", index=False)

# Produce a summary text file with ACTUAL computed values.
summary_lines = [
    "COM7022 Task 1 - reproducible analysis summary",
    f"Source: {DATASET_NAME}",
    f"Original rows: {len(raw)}; modelled USD/unknown-currency rows: {len(model_data)}",
    f"Models compared: {', '.join(comparison['model'])}",
    f"Suggested model pending human review: {suggested_model}",
    f"PCA two-component explained variance: {100*pca_variance.sum():.2f}%",
    "Caution: DBSCAN internal metrics exclude noise; other models use full modelling data.",
    "Caution: sold is cumulative, and available can be censored or missing.",
    "Official submission also requires code appendix, screenshots, and one Word/PDF file."
]
(OUTPUT_DIR / "25_results_summary.txt").write_text("\n".join(summary_lines), encoding="utf-8")

print("Files exported:")
for fp in sorted(OUTPUT_DIR.iterdir()):
    if fp.is_file():
        print(" •", fp.name)

# In Google Colab, this optional block packages every generated output for download.
# Uncomment the final 2 lines if you want the ZIP file immediately.
# import shutil
# shutil.make_archive("COM7022_outputs", "zip", OUTPUT_DIR)
```
